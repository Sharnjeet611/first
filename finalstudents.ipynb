```python
def standardize_lead(x):

    x = str(x).strip().upper()

    if x in ['CAMPUS', 'COLLEGE', 'IG', 'AMBASSADOR','SEMINAR','WEBINAR','FREE']:
        return "CAMPAIGN"

    elif x in ['WEBSITE', 'GOOGLE', 'LANDING','WEB']:
        return "WEBSITE"

    elif x in ['STUDENT', 'REFERRAL', 'REFERRED']:
        return "REFERENCE"

    elif x in ['YOUTUBE', 'WHATSAPP', 'INSTAGRAM', 'FACEBOOK', 'FB','WA','INSTA','LINKEDIN']:
        return "SOCIAL_NETWORKING"
    
    else:
        return "Not Provided"


df["Lead Source"].apply(standardize_lead).unique()
```

```python
df["Lead Source"]=df["Lead Source"].apply(standardize_lead)
```
